```python
monthly_sales = (
    sales
    .groupby(["Month", "Category"])
    .agg(
        Total_Sales=("Amount", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales = monthly_sales.merge(
    sales_target[["Month", "Category", "Target"]],
    on=["Month", "Category"],
    how="left"
)

monthly_sales.head()
```

```python
monthly_sales["Target"].isnull().sum()
```

```python
monthly_sales["Achievement %"] = (
    monthly_sales["Total_Sales"] /
    monthly_sales["Target"]
) * 100

monthly_sales.head()
```

## Data Warehouse

```python
dim_customer = (
    sales[["CustomerName"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_customer["customer_id"] = (
    dim_customer.index + 1
)

dim_customer = dim_customer[
    ["customer_id", "CustomerName"]
]

dim_customer.head()
```

```python
dim_product = (
    sales[["Category", "Sub-Category"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_product["product_id"] = (
    dim_product.index + 1
)

dim_product = dim_product[
    ["product_id", "Category", "Sub-Category"]
]

dim_product.head()
```

```python
dim_product[["product_id", "Category"]].drop_duplicates()
```

```python
dim_category = (
    sales[["Category"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_category["category_id"] = dim_category.index + 1

dim_category
```

```python
dim_location = (
    sales[["State", "City"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_location["location_id"] = (
    dim_location.index + 1
)

dim_location = dim_location[
    ["location_id", "State", "City"]
]

dim_location.head()
```

```python
dim_date = (
    sales[["Order Date"]]
    .drop_duplicates()
    .sort_values("Order Date")
    .reset_index(drop=True)
)

dim_date["date_id"] = (
    dim_date["Order Date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)

dim_date["day"] = dim_date["Order Date"].dt.day

dim_date["month"] = dim_date["Order Date"].dt.month

dim_date["month_name"] = dim_date["Order Date"].dt.strftime("%B")

dim_date["quarter"] = (
    "Q" + dim_date["Order Date"].dt.quarter.astype(str)
)

dim_date["year"] = dim_date["Order Date"].dt.year
```

```python
dim_date = dim_date[
    [
        "date_id",
        "Order Date",
        "day",
        "month",
        "month_name",
        "quarter",
        "year"
    ]
]
```

```python
dim_date.head()
```

```python
fact_sales = sales.merge(
    dim_customer,
    on="CustomerName",
    how="left"
)

fact_sales = fact_sales.merge(
    dim_product,
    on=["Category", "Sub-Category"],
    how="left"
)

fact_sales = fact_sales.merge(
    dim_location,
    on=["State", "City"],
    how="left"
)

fact_sales["date_id"] = (
    fact_sales["Order Date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)
```

```python
fact_sales = fact_sales[
    [
        "Order ID",
        "date_id",
        "customer_id",
        "product_id",
        "location_id",
        "Quantity",
        "Amount",
        "Profit"
    ]
]
```

```python
fact_sales = fact_sales.rename(
    columns={
        "Order ID": "order_id",
        "Quantity": "quantity",
        "Amount": "sales_amount",
        "Profit": "profit"
    }
)
```

```python
fact_sales.head()
```

```python
dim_category = (
    sales[["Category"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_category["category_id"] = (
    dim_category.index + 1
)

fact_sales_target = monthly_sales.merge(
    dim_category,
    on="Category",
    how="left"
)
```

```python
fact_sales_target["target_date"] = (
    fact_sales_target["Month"]
    .dt.to_timestamp()
)

fact_sales_target["date_id"] = (
    fact_sales_target["target_date"]
    .dt.strftime("%Y%m%d")
    .astype(int)
)
```

```python
fact_sales_target = fact_sales_target[
    [
        "date_id",
        "category_id",
        "Target"
    ]
]

fact_sales_target = fact_sales_target.rename(
    columns={
        "Target": "target_amount"
    }
)
```
